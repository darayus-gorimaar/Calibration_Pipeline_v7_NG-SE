```python
treatment_seeking_raster_original, metadata = read_raster(f"{data_path}/{country_code}_treatmentseeking.asc")

nodata = metadata["NODATA_value"]

valid_mask = ((treatment_seeking_raster_original != nodata))

treatment_seeking_raster_normalized = treatment_seeking_raster_original.astype(float).copy()
treatment_seeking_raster_normalized[valid_mask] = treatment_seeking_raster_original[valid_mask] / 100.0
treatment_seeking_raster_normalized[~valid_mask] = nodata   

write_raster(
    raster=treatment_seeking_raster_normalized,
    file=f"{data_path}/{country_code}_treatment_seeking_normalized.asc",
    xllcorner=metadata["xllcorner"],
    yllcorner=metadata["yllcorner"],
    cellsize=metadata["cellsize"],  # preserve nodata areas
    nodata=nodata,
)

info(f"Normalized treatment seeking raster saved to {data_path}/{country_code}_treatment_seeking_normalized.asc\n")
warn(f"Update file name in input_population_bins.yml to use the normalized raster for calibration\n")
```
