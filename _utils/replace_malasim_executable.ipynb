# After pulling new MalaSim executable, replace all binaries

```python
from pathlib import Path
import shutil

from country_config import *
from utils import *
```

```python
malasim_dir = Path("../malasim-v6-SURF")
if not malasim_dir.exists():
    raise FileNotFoundError(f"Missing malasim directory: {malasim_dir}")

bin_path = malasim_dir / "build" / "bin" 
if not bin_path.exists():
    raise FileNotFoundError(f"Missing bin directory: {bin_path}")

mala_sim_file = bin_path / "MalaSim"
if not mala_sim_file.exists():
    raise FileNotFoundError(f"Missing MalaSim file: {mala_sim_file}")
```

```python
current_working_dir = Path.cwd()
print(f"Current working directory: {current_working_dir}")
```

```python
# for all files in current_working_dir, if the file is a MalaSim executable, replace it with the new one (recursively)
for file in current_working_dir.rglob("MalaSim*"):
    if file.is_file():
        print(f"Replacing {file} with {mala_sim_file}")
        shutil.copy(mala_sim_file, file)
```
